2. Дан одномерный массив целых чисел размерности n, заданных случайным образом из интервала от -20 до 20. Если в массиве есть отрицательные элементы, то отсортировать массив по возрастанию, иначе - по убыванию. Реализовать сортировку алгоритмом сортировки вставками.


```python
import random

def insertion_sort(nums,k):
	for i in range(1, len(nums)):
		temp = nums[i]
		j = i - 1
		if not k:
			while (j >= 0 and temp < nums[j]):
				nums[j + 1] = nums[j]
				j = j - 1
		else:
			while (j >= 0 and temp > nums[j]):
				nums[j + 1] = nums[j]
				j = j - 1
		nums[j + 1] = temp
	
n=int(input("n="))

# Создание массива случайных чисел
Array = [random.randint(-20, 20) for _ in range(n)]
print(f"Массив случайных чисел: {Array}")
NegativeNumbers=False	
#Проверка на наличие отрицательных чисел
for N in Array:
	if N<0:
		NegativeNumbers=True
		break
if NegativeNumbers:
	print("В массиве присутствуют отрицательные значения. Сортируем по возрастанию")
	insertion_sort(Array,0)
else:
	print("В массиве отсутствуют отрицательные значения. Сортируем по убыванию")
	insertion_sort(Array,1)	
print(f"Сортированный массив : {Array}")
```
